```python
#tarea 1 - parte 5
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Cargar los datos desde el archivo CSV
df = pd.read_csv("datos_ambiental_evaluacion_tema9.csv")

# 2. Seleccionar solo las columnas numéricas para la correlación
# (ignoramos 'estacion' y 'tipo_zona' porque no son números)
df_numerico = df.select_dtypes(include=[np.number])

# 3. Calcular la matriz de correlación de Pearson
corr = df_numerico.corr()
print("Matriz de Correlación:")
print(corr.round(4))
```

```python
# 4. Construir el heatmap con seaborn
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0,
            vmin=-1, vmax=1, square=True, ax=ax,
            cbar_kws={"label": "Correlación de Pearson"})

plt.title("Matriz de Correlación - Datos Ambientales")
plt.tight_layout()
plt.savefig("heatmap_seaborn_ambiental.png")
plt.show()
```

```python
# Completa: mascara y segundo heatmap
mascara = np.triu(np.ones_like(corr, dtype=bool))
print("Celdas visibles:", (~mascara).sum())

fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0,
            vmin=-1, vmax=1, square=True, mask=mascara, ax=ax,
            cbar_kws={"label": "Correlación de Pearson"})

plt.title("Matriz de Correlación - Con Máscara")
plt.tight_layout()
plt.savefig("heatmap_seaborn_mascara.png")
plt.show()
```
