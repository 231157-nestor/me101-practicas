```python
from skimage.io import imread
from skimage.filters import gaussian, threshold_otsu
from skimage.color import rgb2gray # 1. Importar la función de conversión a grises
import numpy as np

# Cargar imagen
img = imread('images.png')

# 2. Convertir la imagen a escala de grises
gray_img = rgb2gray(img)

# 3. Aplicar filtro Gaussiano (ya no necesitas channel_axis=-1 porque es de 1 canal)
blur = gaussian(gray_img, sigma=2.5)

# Binarización automática por umbral de Otsu
thresh_val = threshold_otsu(blur)
bin_img = blur > thresh_val
```
