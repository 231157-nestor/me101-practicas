```python
#TAREA 1
from PIL import Image, ImageDraw

# Completa: crea tu imagen con fondo y al menos 2 figuras
img = Image.new("RGB", (120, 120), color=(100, 50, 150)) # Fondo morado
draw = ImageDraw.Draw(img)
draw.rectangle([10, 10, 50, 50], fill=(0, 255, 255))     # figura 1: cuadrado cian
draw.ellipse([60, 60, 110, 110], fill=(255, 0, 128))     # figura 2: círculo rosa

print(img.size, img.mode)
img.save("mi_imagen.png")
```

```python
# Completa: reabrir y verificar pixeles clave
img2 = Image.open("mi_imagen.png")

print("Formato:", img2.format)

# Anota estas coordenadas para la Tarea 3:
print("Píxel de fondo (5, 5):", img2.getpixel((5, 5)))
print("Píxel figura 1 (30, 30):", img2.getpixel((30, 30)))
print("Píxel figura 2 (85, 85):", img2.getpixel((85, 85)))
```

```python
# Completa: las cuatro transformaciones
img_gris = img2.convert("L")
img_rotada = img2.rotate(90, expand=True)
img_mitad = img2.resize((60, 60)) # La mitad exacta de 120x120
img_volteada = img2.transpose(Image.FLIP_TOP_BOTTOM)

print("Tamaño gris:", img_gris.size)
print("Tamaño rotada:", img_rotada.size)
print("Tamaño a la mitad:", img_mitad.size)
print("Tamaño volteada:", img_volteada.size)
```

```python
#TAREA 2
from PIL import ImageEnhance, ImageFilter

# Completa: dos factores de brillo distintos
realzador = ImageEnhance.Brightness(img2)
img_oscura = realzador.enhance(0.5)    # Factor menor a 1.0
img_brillante = realzador.enhance(2.5) # Factor mayor a 2.0

# Verificación de recorte en el píxel de la figura 1 (cuadrado cian)
print("Figura 1 original (30, 30):", img2.getpixel((30, 30)))
print("Figura 1 con realce > 2.0 (30, 30):", img_brillante.getpixel((30, 30)))
# Notarás que los valores originales (0, 255, 255) se recortan y no superan el 255.
```

```python
# Completa: aplica y guarda ambos filtros
img_blur = img2.filter(ImageFilter.GaussianBlur(radius=4))
img_blur.save("mi_imagen_blur.png")

img_bordes = img2.filter(ImageFilter.FIND_EDGES)
img_bordes.save("mi_imagen_bordes.png")
```

```python
import numpy as np
from skimage import color, filters

# Completa: array, rgb2gray, comparacion con convert('L')
arr = np.array(img2)
gris_sk = color.rgb2gray(arr)
gris_pil = np.array(img_gris) # 'img_gris' fue creada en la Tarea 1

# Comparamos el MISMO píxel de fondo anotado en la Tarea 1: (5, 5)
# Nota: En NumPy el acceso es [fila, columna], es decir [Y, X]. 
# Como las coordenadas son idénticas (5, 5), se escribe igual.
print("Pillow convert('L'), fondo [5, 5]:", gris_pil[5, 5])
print("skimage rgb2gray, fondo [5, 5]:", round(gris_sk[5, 5] * 255, 2))
```

```python
# Completa: threshold_otsu y binarizacion
umbral_otsu = filters.threshold_otsu(gris_sk)
binaria = gris_sk > umbral_otsu

print("Umbral de Otsu calculado automáticamente:", round(umbral_otsu, 4))
print("Proporción de píxeles 'True' (sobre el umbral):", round(binaria.mean(), 4))
```
