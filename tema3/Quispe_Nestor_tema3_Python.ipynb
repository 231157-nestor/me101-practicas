```python
import numpy as np
notas = np.array([11,14,20,8,16,13])
print("Shape:", notas.shape)
print("Dtype:", notas.dtype)
print("Ndim:", notas.ndim)
```

```python
# Completa: promedio, maximo, y cantidad de aprobados
promedio = notas.mean()
maximo = notas.max()
cantidad_aprobados = (notas >= 10.5).sum()  # pista: suma un array booleano

print("Promedio:", promedio)
print("Maximo:", maximo)
print("Aprobados:", cantidad_aprobados)
```

```python
import pandas as pd

# Completa: crea el diccionario de datos y el DataFrame
datos = {
    "nombre": ["Ana", "Luis", "Carlos", "María", "Jorge"],
    "nota": [15, 9, 18, 12, 8],
    "asistencia_pct": [85, 90, 65, 75, 50],
}
df = pd.DataFrame(datos)

# Completa: filtra estudiantes con nota >= 10.5 Y asistencia_pct >= 70
# pista: combina dos condiciones booleanas con el operador &
estudiantes_aprobados = df[(df["nota"] >= 10.5) & (df["asistencia_pct"] >= 70)]
print(estudiantes_aprobados)
```
