```python
import numpy as np
import pandas as pd
from scipy import stats

df = pd.read_csv("tema4/estudiantes_limpio.csv")

# Cálculo de estadísticos
media = df["nota"].mean()
mediana = df["nota"].median()
std_muestral = df["nota"].std(ddof=1)  # pandas usa ddof=1 por defecto
var_poblacional = np.var(df["nota"], ddof=0)  # numpy usa ddof=0 por defecto
q1 = df["nota"].quantile(0.25)
q3 = df["nota"].quantile(0.75)
iqr = q3 - q1  # o stats.iqr(df["nota"])

print(f"Media: {media:.4f}")
print(f"Mediana: {mediana:.4f}")
print(f"Desv. Estándar Muestral (pandas): {std_muestral:.4f}")
print(f"Varianza Poblacional (numpy): {var_poblacional:.4f}")
print(f"Q1: {q1:.4f}, Q3: {q3:.4f}, IQR: {iqr:.4f}")
```

```python
# Con pandas (sesgo muestral corregido, exceso de curtosis de Fisher)
skew_pandas = df["nota"].skew()
kurt_pandas = df["nota"].kurt()

# Con scipy.stats (por defecto bias=True / no corregido por tamaño muestral)
skew_scipy = stats.skew(df["nota"])
kurt_scipy = stats.kurtosis(df["nota"])  # Exceso de curtosis (normal = 0)

print(f"Skewness (pandas): {skew_pandas:.4f} | Skewness (scipy): {skew_scipy:.4f}")
print(f"Kurtosis (pandas): {kurt_pandas:.4f} | Kurtosis (scipy): {kurt_scipy:.4f}")
```

```python
def resumen_estadistico(serie, decimales=4):
    n = len(serie.dropna())
    media = serie.mean()
    mediana = serie.median()
    desv_std = serie.std(ddof=1)
    cv_pct = (desv_std / media) * 100 if media != 0 else np.nan

    return {
        "n": n,
        "media": round(media, decimales),
        "mediana": round(mediana, decimales),
        "desv_std": round(desv_std, decimales),
        "cv_pct": round(cv_pct, decimales),
    }


print(resumen_estadistico(df["nota"]))
```

```python
def clasificar_dispersion(cv_pct):
    if cv_pct < 15:
        return "Baja"
    elif 15 <= cv_pct < 30:
        return "Moderada"
    else:
        return "Alta"


columnas = ["nota", "asistencia_pct"]
for columna in columnas:
    res = resumen_estadistico(df[columna])
    nivel = clasificar_dispersion(res["cv_pct"])
    print(f"Columna: {columna}")
    print(f"  Resumen: {res}")
    print(f"  Dispersión: {nivel} (CV = {res['cv_pct']}%)\n")
```
