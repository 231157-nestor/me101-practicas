```python
#ejercicio 1
import numpy as np
import pandas as pd

datos = {
    "nombre": ["Nestor", "Carlos", "Jose", "David", "elva", "Fernando"],
    "curso": ["ME101", "ME101", "ME102", "ME101", "ME102", "ME102"],
    "nota": [16.5, np.nan, 14.0, 11.5, 18.0, 15.0],  
    "asistencia_pct": [92.0, 85.0, np.nan, 78.0, 95.0, 88.0]  
}

df_original = pd.DataFrame(datos)
df_original.to_csv("estudiantes.csv", index=False)
```

```python
#ejercicio 2
df = pd.read_csv("estudiantes.csv")

print("Forma del DataFrame (filas, columnas):", df.shape)
print("\nValores faltantes por columna:")
print(df.isna().sum())
```

```python
#ejercicio 3
df_limpio = df.dropna(subset=["nota"]).copy()

# Rellenar asistencia_pct faltante con el promedio de esa columna
promedio_asistencia = df_limpio["asistencia_pct"].mean()
df_limpio["asistencia_pct"] = df_limpio["asistencia_pct"].fillna(promedio_asistencia)

# Verificar que no queden valores faltantes
print("Valores faltantes restantes:")
print(df_limpio.isna().sum())
```

```python
#ejercicio 4
promedio_por_curso = df_limpio.groupby("curso")["nota"].mean()
print("Promedio de nota por curso:")
print(promedio_por_curso)

df_limpio.to_csv("estudiantes_limpio.csv", index=False)
```
